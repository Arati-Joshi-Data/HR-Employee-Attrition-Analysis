```python
# Import required libraries
import pandas as pd
import numpy as np

# Upload the HR dataset
from google.colab import files
uploaded = files.upload()

# Read the uploaded CSV file
import io
file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

# Correct the first column name if required
df.rename(columns={'na': 'Age'}, inplace=True)

# Display dataset details
print("Dataset Shape:", df.shape)
print("\nFirst 5 Records:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# 1. Check dataset information
print("Dataset Information:")
df.info()

# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# 4. Check statistical summary
print("\nStatistical Summary:")
display(df.describe(include='all').T)
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Employee Attrition Distribution

print("Attrition Count:")
print(df['Attrition'].value_counts())

print("\nAttrition Percentage:")
print(df['Attrition'].value_counts(normalize=True) * 100)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Attrition')

plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')

plt.show()
```

```python
# Attrition analysis by different employee attributes

attributes = [
    'Department',
    'JobRole',
    'OverTime',
    'BusinessTravel',
    'Gender',
    'MaritalStatus'
]

for col in attributes:

    print(f"\nAttrition by {col}")

    attrition_rate = pd.crosstab(
        df[col],
        df['Attrition'],
        normalize='index'
    ) * 100

    display(attrition_rate.round(2))

    attrition_rate.plot(
        kind='bar',
        stacked=True,
        figsize=(9, 5)
    )

    plt.title(f'Attrition Percentage by {col}')
    plt.xlabel(col)
    plt.ylabel('Percentage')
    plt.legend(title='Attrition')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Numerical analysis based on employee attrition

numerical_columns = [
    'Age',
    'MonthlyIncome',
    'TotalWorkingYears',
    'YearsAtCompany',
    'DistanceFromHome'
]

for col in numerical_columns:

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x='Attrition',
        y=col
    )

    plt.title(f'{col} Distribution by Attrition')
    plt.xlabel('Attrition')
    plt.ylabel(col)

    plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns for correlation analysis
numerical_columns = [
    'Age',
    'MonthlyIncome',
    'TotalWorkingYears',
    'YearsAtCompany',
    'DistanceFromHome'
]

# Convert Attrition into numerical format
# Yes = 1, No = 0
df['Attrition_Numeric'] = df['Attrition'].map({
    'Yes': 1,
    'No': 0
})

# Create correlation matrix
correlation_data = df[numerical_columns + ['Attrition_Numeric']]

correlation_matrix = correlation_data.corr()

# Display correlation matrix
print("Correlation Matrix:")
display(correlation_matrix)

# Visualize correlation using heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Analysis - Employee Attrition')
plt.show()
```

```python

# Overall Employee Attrition Summary

total_employees = len(df)

attrition_count = (df['Attrition'] == 'Yes').sum()
retained_count = (df['Attrition'] == 'No').sum()

attrition_rate = (attrition_count / total_employees) * 100
retention_rate = (retained_count / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Who Left:", attrition_count)
print("Employees Retained:", retained_count)
print("Attrition Rate: {:.2f}%".format(attrition_rate))
print("Retention Rate: {:.2f}%".format(retention_rate))
```
